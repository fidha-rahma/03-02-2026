```python
# 1. Data Loading
import pandas as pd
import numpy as np

df = pd.read_csv("dataset_sensor_iot_kelas.csv")
df.head()
```

```python
# 2. Data Inspection
df.head()
df.info()
df.describe()
df.shape
df.isnull().sum()
df.duplicated().sum
```

```python
# 3. Data Cleansing :
# Membersikan kolom suhu
df['suhu_celsius'] = pd.to_numeric(
    df['suhu_celsius']
    .astype(str)
    .str.extract(r"([-+]?\d+(?:\.\d+)?)")[0],
    errors='coerce'
)
#  Membersihkan kolom kelembapan
df['kelembapan_persen'] = pd.to_numeric(
    df['kelembapan_persen'],
    errors='coerce'
)
# Menangani nilai tidak masuk akal
df.loc[-df["suhu_celsius"].between(15, 40), "suhu_celsius"] = np.nan
df.loc[-df["kelembapan_persen"].between(0, 100), "kelembapan_persen"] = np.nan
# Menangani missing value
df['suhu_celsius'] = df['suhu_celsius'].fillna(df['suhu_celsius'].median())
df['kelembapan_persen'] = df['kelembapan_persen'].fillna(df['kelembapan_persen'].median())

df["status_sensor"] = df["status_sensor"].fillna(df["status_sensor"].mode()[0])

# Menangani data duplikat
df = df.drop_duplicates()
#  Memastikan tipe data
df.info()
```

```python
# 4. Data Manipulaation

#1. Filtering
df_panas = df[df["suhu_celsius"] >= 30]
df_panas

#2. Sorting
df_sorted = df.sort_values(by="suhu_celsius", ascending=False)
df_sorted.head()

#3. Membuat kolom turunan
def kategori_suhu(suhu):
    if suhu < 20:
      return "Dingin"
    elif suhu < 30:
      return "Normal"
    else:
      return "Panas"
df["kategori_suhu"] = df["suhu_celsius"].apply(kategori_suhu)
df.head()

#4. Groupby dan Agregasi
ringkasan_ruang = df.groupby("ruang").agg(
    rata_rata_suhu=("suhu_celsius", "mean"),
    rata_rata_kelembapan=("kelembapan_persen", "mean"),
    jumlah_pembacaan=("suhu_celsius", "count")
    ).reset_index()
ringkasan_ruang
```

```python
# 5. Hasil cleaning
print("Jumlah missing value:")
print(df.isnull().sum())

print("\nJumlah duplikat:")
print(df.duplicated().sum())

print("\nUkuran dataset setelah cleaning:")
print(df.shape)
```
