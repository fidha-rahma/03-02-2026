```
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Tugas Analisis 1:
Bandingkan hasil harga * 0.9 di atas dengan jika harga berupa list Python
biasa (bukan array). Apa yang terjadi bila kamu mencoba [5000,7000,3000] * 0.9? Jelaskan
mengapa berbeda.

Jawaban:
Jika menggunakan NumPy array, perkalian harga * 0.9 akan menghitung setiap elemen dengan angka 0.9 sehingga menghasilkan diskon 10%. Namun, jika menggunakan list Python biasa, perkalian list dengan angka desimal seperti [5000,7000,3000] * 0.9 akan menghasilkan error karena list hanya bisa dikalikan dengan bilangan bulat untuk mengulang isi list, bukan untuk menghitung setiap elemennya.

```
df = pd.read_csv('dataset_penjualan_kantin (1).csv')

print(df.head())        # 5 baris pertama
print(df.info())        # tipe data & jumlah non-null tiap kolom
print(df.describe())    # statistik ringkas kolom numerik
print(df.shape)         # jumlah (baris, kolom)
```

Tugas Analisis 3:
Dari hasil df.info(), kolom mana yang jumlahnya non-null-nya lebih sedikit dari julah baris total? Apa artinya?

Jawaban:
Kolom yang jumlah non-null-nya lebih sedikit dari jumlah baris total adalah:

- harga_satuan sebanyak 66 dari 69 baris.
- nama_kasir sebanyak 66 dari 69 baris.

kolom-kolom memiliki data kosong atau missing value, masih ada beberapa informasi transaksi yang belum terisi lengkap.

```
df.columns = df.columns.str.strip().str.lower()

print(df.columns)
print(df.isnull().sum())

df['terjual'] = df['terjual'].fillna(0)
df = df.dropna(subset=['menu'])
```

Tugas Analisis 4:
Mengapa pada contoh di atas kolom terjual diisi (fillna) sedangkan baris
dengan menu kosong justru dihapus (dropna)? Diskusikan alasan logisnya.

Jawaban:
Kolom terjual diisi dengan angka 0 karena data yang kosong dianggap tidak memiliki jumlah penjualan. Sedangkan baris dengan menu kosong dihapus karena nama menu merupakan informasi penting untuk mengetahui produk yang dijual. Jika nama menu tidak ada, data tersebut akan sulit dianalisis.

```
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

```
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting

df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan

ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```
