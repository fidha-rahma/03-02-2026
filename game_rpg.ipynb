```python
class Hero:
    # Constructor
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    # Method info
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang
    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    # Method diserang
    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")


# Class Mage adalah anak dari class Hero
class Mage(Hero):

    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(name, hp, attack_power)
        self.mana = mana

    # Override method info
    def info(self):
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")


# -- Main Program Baru --

print("\n--- Update Class Hero ---")

eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()

eudora.serang(balmond)          # Serangan biasa
eudora.skill_fireball(balmond)  # Skill khusus Mage


# -- Main Program --

hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

# Mengubah HP Layla menjadi 500
hero1.hp = 500

# Menampilkan informasi hero
hero1.info()
hero2.info()

# Pertarungan
print("\n--- Pertarungan Dimulai ---")

hero1.serang(hero2)
hero2.serang(hero1)
```

```python
from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.

class GameUnit(ABC):

      @abstractmethod
      def serang(self, target):
          pass

      @abstractmethod
      def info(self):
          pass
# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

 # Kita WAJIB membuat method serang, kalau tidak akan Error
    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")
    def info(self):
        print(f"Saya adalah Hero: {self.nama}")
class Monster(GameUnit):
  def __init__(self, jenis):
    self.jenis = jenis
 # Implementasi serang versi Monster
  def serang(self, target):
      print(f"Monster {self.jenis} menggigit {target}!")
  def info(self):
      print(f"Saya adalah Monster: {self.jenis}")
# -- Uji Coba --
# unit = GameUnit() # ERROR! Abstract class tidak bisa jadi objek.
h = Hero("Alucard")
m = Monster("Serigala")
h.info()
m.info()
```

Meskipun tidak bisa dibuat menjadi objek secara langsung, GameUnit berguna untuk menjadi dasar bagi class lain, seperti Hero dan Monster. Dengan begitu, setiap class turunan wajib memiliki method serang() dan info().

```python
# Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")


# Child Class 1
class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")


# Child Class 2
class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")


# Child Class 3
class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")


# -- Penerapan Polymorphism --
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord")
]

print("--- PERANG DIMULAI ---")

for pahlawan in pasukan:
    pahlawan.serang()
```

2. Apa yang terjadi?
Program tidak akan menjalankan tembak_panah(). Karena Archer tidak lagi memiliki method serang(), Python akan menggunakan method serang() milik parent class Hero. Akibatnya, ketika Miya diproses, hasilnya menjadi: Hero menyerang dengan tangan kosong.
Mengapa nama method harus sama?
Karena polimorfisme bekerja dengan memberikan perintah yang sama, yaitu serang(), kepada objek yang berbeda. Setiap child class kemudian memberikan perilaku yang berbeda untuk method tersebut
Jadi:
Mage → serang() → Bola Api
Archer → serang() → Memanah
Fighter → serang() → Pedang
Healer → serang() → Menyembuhkan
Jika namanya berbeda seperti tembak_panah(), maka perintah pahlawan.serang() tidak akan memanggil perilaku khusus milik Archer.

```python
from abc import ABC, abstractmethod


# ==========================================
# 1. ABSTRACTION
# ==========================================

class KamarHotel(ABC):
    def __init__(self, nama_kamar, harga_dasar, stok):
        self.nama_kamar = nama_kamar
        self.__harga_dasar = harga_dasar
        self.__stok = stok

    # Getter untuk melihat harga dasar
    def get_harga_dasar(self):
        return self.__harga_dasar

    # Getter untuk melihat stok
    def get_stok(self):
        return self.__stok

    # Method untuk menambah stok
    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print(
                f"Gagal update stok {self.nama_kamar}! "
                f"Stok tidak boleh negatif ({jumlah})."
            )
        else:
            self.__stok += jumlah
            print(
                f"Berhasil menambahkan stok {self.nama_kamar}: "
                f"{jumlah} unit."
            )

    # Abstract Method
    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass


# ==========================================
# 2. INHERITANCE
# ==========================================

class KamarDeluxe(KamarHotel):
    def __init__(self, nama_kamar, harga_dasar, stok, fasilitas):
        super().__init__(nama_kamar, harga_dasar, stok)
        self.fasilitas = fasilitas

    # POLYMORPHISM - Override
    def tampilkan_detail(self):
        print(
            f"[DELUXE] {self.nama_kamar} | "
            f"Fasilitas: {self.fasilitas}"
        )

    # POLYMORPHISM - Override
    def hitung_harga_total(self, jumlah_malam):
        harga = self.get_harga_dasar()

        # Pajak Deluxe = 10%
        pajak = harga * 10 / 100

        subtotal = (harga + pajak) * jumlah_malam

        print(f"Harga Dasar/Malam: Rp {harga:,.0f}")
        print(f"Pajak (10%): Rp {pajak:,.0f}")
        print(f"Menginap: {jumlah_malam} malam")
        print(f"Subtotal: Rp {subtotal:,.0f}")

        return subtotal


class KamarStandard(KamarHotel):
    def __init__(self, nama_kamar, harga_dasar, stok, kapasitas):
        super().__init__(nama_kamar, harga_dasar, stok)
        self.kapasitas = kapasitas

    # POLYMORPHISM - Override
    def tampilkan_detail(self):
        print(
            f"[STANDARD] {self.nama_kamar} | "
            f"Kapasitas: {self.kapasitas}"
        )

    # POLYMORPHISM - Override
    def hitung_harga_total(self, jumlah_malam):
        harga = self.get_harga_dasar()

        # Pajak Standard = 5%
        pajak = harga * 5 / 100

        subtotal = (harga + pajak) * jumlah_malam

        print(f"Harga Dasar/Malam: Rp {harga:,.0f}")
        print(f"Pajak (5%): Rp {pajak:,.0f}")
        print(f"Menginap: {jumlah_malam} malam")
        print(f"Subtotal: Rp {subtotal:,.0f}")

        return subtotal


# ==========================================
# 3. POLYMORPHISM
# ==========================================

def proses_transaksi(daftar_pesanan):
    total_tagihan = 0

    print("\n--- STRUK PEMESANAN ---")

    for nomor, (kamar, jumlah_malam) in enumerate(
        daftar_pesanan, start=1
    ):
        print(f"{nomor}. ", end="")

        # Memanggil method yang sama
        # tetapi hasilnya berbeda sesuai objek
        kamar.tampilkan_detail()

        total_tagihan += kamar.hitung_harga_total(
            jumlah_malam
        )

    print("----------------------------------------")
    print(f"TOTAL TAGIHAN: Rp {total_tagihan:,.0f}")


# ==========================================
# 4. PROGRAM UTAMA
# ==========================================

print("--- SETUP DATA KAMAR ---")

# Admin membuat 1 kamar Deluxe
kamar_deluxe = KamarDeluxe(
    "Kamar Deluxe Sea View",
    1_500_000,
    0,
    "Private Pool"
)

# Admin membuat 1 kamar Standard
kamar_standard = KamarStandard(
    "Kamar Standard Superior",
    500_000,
    0,
    "2 Orang"
)


# ==========================================
# 5. UPDATE STOK
# ==========================================

# Menambahkan stok Deluxe
kamar_deluxe.tambah_stok(10)

# Mencoba memasukkan stok negatif
kamar_standard.tambah_stok(-5)

# Menambahkan stok Standard dengan benar
kamar_standard.tambah_stok(20)


# ==========================================
# 6. PEMESANAN
# ==========================================

daftar_pesanan = [
    (kamar_deluxe, 2),
    (kamar_standard, 1)
]

proses_transaksi(daftar_pesanan)
```
